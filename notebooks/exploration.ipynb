```python
import pandas as pd
```

```python
data = pd.read_csv('../data/gender.csv')
print(data.head())
print(len(data))
```

```python
def analyze_author_distribution(df):
    # Get post counts per author
    author_counts = df['auhtor_ID'].value_counts()

    # Calculate key statistics
    total_authors = len(author_counts)
    total_posts = author_counts.sum()
    mean_posts = author_counts.mean()
    median_posts = author_counts.median()
    max_posts = author_counts.max()
    min_posts = author_counts.min()

    # Calculate percentiles
    percentiles = author_counts.quantile([0.25, 0.75, 0.9, 0.95, 0.99])

    print(f"Author Post Distribution Statistics:")
    print(f"Total Authors: {total_authors:,}")
    print(f"Total Posts: {total_posts:,}")
    print(f"\nPosts per Author:")
    print(f"Mean: {mean_posts:.1f}")
    print(f"Median: {median_posts:.1f}")
    print(f"Min: {min_posts}")
    print(f"Max: {max_posts}")
    print(f"\nPercentiles:")
    print(f"25th percentile: {percentiles[0.25]:.1f} posts")
    print(f"75th percentile: {percentiles[0.75]:.1f} posts")
    print(f"90th percentile: {percentiles[0.90]:.1f} posts")
    print(f"95th percentile: {percentiles[0.95]:.1f} posts")
    print(f"99th percentile: {percentiles[0.99]:.1f} posts")

    # Calculate proportion of authors with different post counts
    counts_1_5 = sum((author_counts >= 1) & (author_counts <= 5))
    counts_6_20 = sum((author_counts > 5) & (author_counts <= 20))
    counts_21_100 = sum((author_counts > 20) & (author_counts <= 100))
    counts_over_100 = sum(author_counts > 100)

    print(f"\nAuthor Distribution:")
    print(f"Authors with 1-5 posts: {counts_1_5:,} ({counts_1_5/total_authors*100:.1f}%)")
    print(f"Authors with 6-20 posts: {counts_6_20:,} ({counts_6_20/total_authors*100:.1f}%)")
    print(f"Authors with 21-100 posts: {counts_21_100:,} ({counts_21_100/total_authors*100:.1f}%)")
    print(f"Authors with >100 posts: {counts_over_100:,} ({counts_over_100/total_authors*100:.1f}%)")
```

```python
analyze_author_distribution(data)
```
