```python
import pandas as pd
```

```python
df = pd.read_csv("../data/gender.csv")
```

```python
import nltk
from nltk.corpus import words

# Download the word list if not already downloaded
nltk.download('words')
```

```python
english_vocab = set(words.words())

# Function to check for non-English words
def contains_non_english(text):
    words_in_text = set(text.split())  # Split text into words
    non_english_words = [word for word in words_in_text if word.lower() not in english_vocab]
    return len(non_english_words) > 0

# Apply to your DataFrame
df['contains_non_english'] = df['post'].apply(contains_non_english)

# Count the number of posts with non-English words
non_english_count = df[df['contains_non_english']].shape[0]
print(f"Number of posts with non-English words: {non_english_count}")
```

```python
from langdetect import detect

df['language'] = df['post'].apply(detect)

language_counts_absolute = df['language'].value_counts()

# Print the result
print(language_counts_absolute)
```

```python
from langdetect import detect_langs

df2 = df.sample(1000)

def detect_language_with_confidence(text):
    langs = detect_langs(text)
    # Extract the most probable language and its confidence score
    most_probable_lang = max(langs, key=lambda x: x.prob)
    return most_probable_lang.lang, most_probable_lang.prob

df2['language'], df2['confidence'] = zip(*df2['post'].apply(detect_language_with_confidence))
non_english_posts = df2[df2['language'] != 'en']

for index, row in non_english_posts.iterrows():
    print(f"Post {index}: (Language: {row['language']}, Confidence: {row['confidence']:.4f}),{row['post']}")

```
